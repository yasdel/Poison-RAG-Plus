```python
####################################
# A) OpenAI Embedding Extraction
####################################
import os
import io
import gzip
import math
import random
import time
import pandas as pd
import concurrent.futures
from tqdm import tqdm
from typing import List, Optional
from google.colab import files
import openai
import shutil

openai.api_key = os.getenv("OPENAI_API_KEY")  # Or manually set: openai.api_key = "your-key"



# 5. HELPER FUNCTIONS

def create_combined_text(row, column_name: str) -> str:
    """
    Combine title, genres, and column text (tags) into a single string.
    """
    title = row.get('title', '')
    genre = row.get('genres', '').replace('|', ' ') if 'genres' in row else ''
    tags = row.get(column_name, [])
    if isinstance(tags, str):
        try:
            tags = eval(tags)  # if it's a string of list-likes
        except:
            pass
    tags_str = ' '.join(tags) if isinstance(tags, list) else str(tags)
    return f"{title} {genre} {tags_str}"

def generate_openai_embedding(text: str, model: str = "text-embedding-ada-002") -> Optional[List[float]]:
    """
    Generate a single embedding from OpenAI's API.
    """
    if not text.strip():
        return []
    try:
        response = openai.Embedding.create(input=text, model=model)
        embedding = response['data'][0]['embedding']
        # Small random pause to reduce risk of hitting rate limits
        time.sleep(random.uniform(0.3, 0.5))
        return embedding
    except Exception as e:
        print(f"Error generating embedding for text: {text[:50]}... \nError: {e}")
        return []

def get_openai_embeddings_parallel(texts: List[str], max_workers: int = 10) -> List[List[float]]:
    """
    Generate embeddings for a list of texts using ThreadPoolExecutor.
    """
    embeddings = []
    with concurrent.futures.ThreadPoolExecutor(max_workers=max_workers) as executor:
        future_to_text = {executor.submit(generate_openai_embedding, txt): txt for txt in texts}
        for future in tqdm(concurrent.futures.as_completed(future_to_text),
                           total=len(texts), desc="Generating OpenAI Embeddings"):
            emb = future.result()
            embeddings.append(emb)
    return embeddings

def estimate_chunk_size(df: pd.DataFrame, sample_size: int = 1000, max_size_mb: int = 25) -> int:
    """
    Estimate how many rows per chunk to keep each file ~25MB max when gzipped.
    """
    max_size_bytes = max_size_mb * 1024 * 1024
    sample = df.head(sample_size)
    csv_buffer = io.StringIO()
    sample.to_csv(csv_buffer, index=False)
    sample_size_bytes = len(csv_buffer.getvalue().encode('utf-8'))
    avg_size_per_row = sample_size_bytes / max(1, len(sample))
    estimated_rows = max(1, math.floor(max_size_bytes / avg_size_per_row))
    return estimated_rows

def split_dataframe(df: pd.DataFrame, chunk_size: int):
    """
    Split the DataFrame into multiple DataFrames each up to 'chunk_size' rows.
    """
    num_chunks = math.ceil(len(df) / chunk_size)
    return [df.iloc[i*chunk_size:(i+1)*chunk_size] for i in range(num_chunks)]

# 6. RUN THE OPENAI EMBEDDING EXTRACTION
max_workers = 12
output_dir = f"openai_embeddings_attacked_ds_name_{ds_name}_{flag_case}"
os.makedirs(output_dir, exist_ok=True)

if embedding_method.lower() == 'openai':
  for column in embedding_columns:
      print(f"\nProcessing embeddings for column: {column}")

      # Create combined text
      df[f'combined_{column}'] = df.apply(lambda row: create_combined_text(row, column), axis=1)
      texts = df[f'combined_{column}'].tolist()

      # Generate embeddings in parallel
      embeddings = get_openai_embeddings_parallel(texts, max_workers=max_workers)

      # Build a DataFrame to store embeddings
      embedding_df = pd.DataFrame({
          'itemId': df['itemId'],
          'embeddings': embeddings
      })

      # Convert vectors to space-separated strings
      embedding_df['embeddings'] = embedding_df['embeddings'].apply(lambda x: ' '.join(map(str, x)) if x else '')

      # Shuffle rows for random distribution
      embedding_df = embedding_df.sample(frac=1, random_state=42).reset_index(drop=True)

      # Split into chunks
      chunk_size = estimate_chunk_size(embedding_df, sample_size=1000, max_size_mb=25)
      chunks = split_dataframe(embedding_df, chunk_size)

      # Save each chunk
      for idx, chunk in enumerate(chunks):
          output_file = os.path.join(output_dir, f"openai_{column}_part{idx+1}.csv.gz")
          with gzip.open(output_file, 'wt', encoding='utf-8') as f:
              chunk.to_csv(f, index=False)
          print(f"Saved: {output_file}")

  print("\nOpenAI Embedding Extraction Complete!")


  # Zip and download the folder
  zip_file_name = f"{output_dir}.zip"
  shutil.make_archive(output_dir, 'zip', output_dir)
  files.download(zip_file_name)
```

```python
if embedding_method.lower() == 'llama3':
  #################################################
  # C) Llama Embedding Extraction (Hugging Face)
  #################################################
  !pip install --quiet transformers accelerate bitsandbytes huggingface_hub
  import os
  import io
  import gzip
  import math
  import torch
  import pandas as pd
  import concurrent.futures
  from tqdm import tqdm
  from transformers import AutoTokenizer, AutoModelForCausalLM
  from huggingface_hub import login
  from google.colab import files
  import shutil

  # 1. GPU/CPU selection
  device = "cuda" if torch.cuda.is_available() else "cpu"
  model_dtype = torch.float16 if torch.cuda.is_available() else torch.float32

  # 2. OPTIONAL: Rename Columns or Inspect
  if 'artist' in df.columns:
      df.rename(columns={'artist': 'title'}, inplace=True)
  print("Dataframe columns:", df.columns)
  display(df.head())

  login(hf_token)

  model_name = "meta-llama/Llama-2-7b-hf"  # Example Llama2 model name
  tokenizer = AutoTokenizer.from_pretrained(model_name, use_auth_token=hf_token)

  # Use 'device_map="auto"' only if using a GPU; otherwise None
  model = AutoModelForCausalLM.from_pretrained(
      model_name,
      device_map="auto" if torch.cuda.is_available() else None,
      torch_dtype=model_dtype,
      use_auth_token=hf_token
  )
  model.eval()  # Put model in evaluation mode

# 5. HELPER FUNCTIONS
def create_combined_text(row, column_name: str) -> str:
    title = row.get('title', '')
    genre = row.get('genres', '').replace('|', ' ') if 'genres' in row else ''
    tags = row.get(column_name, [])
    if isinstance(tags, str):
        try:
            tags = eval(tags)
        except:
            pass
    tags_str = ' '.join(tags) if isinstance(tags, list) else str(tags)
    return f"{title} {genre} {tags_str}"

def get_llama_embeddings(batch_texts):
    """
    Embeds a batch of texts using Llama model's last hidden state (mean-pooled).
    """
    if tokenizer.pad_token is None:
        # Some Llama-based models might not have a pad token
        tokenizer.pad_token = tokenizer.eos_token
        model.resize_token_embeddings(len(tokenizer))

    inputs = tokenizer(
        batch_texts,
        return_tensors="pt",
        padding=True,
        truncation=True
    ).to(device)

    with torch.no_grad():
        outputs = model(**inputs, output_hidden_states=True)
        last_hidden = outputs.hidden_states[-1]  # Final layer
        embeddings = last_hidden.mean(dim=1)

    return embeddings.cpu().numpy().tolist()

def estimate_chunk_size(df: pd.DataFrame, sample_size: int = 1000, max_size_mb: int = 25) -> int:
    max_size_bytes = max_size_mb * 1024 * 1024
    sample = df.head(sample_size)
    csv_buffer = io.StringIO()
    sample.to_csv(csv_buffer, index=False)
    sample_size_bytes = len(csv_buffer.getvalue().encode('utf-8'))
    avg_size_per_row = sample_size_bytes / max(1, len(sample))
    return max(1, math.floor(max_size_bytes / avg_size_per_row))

def split_dataframe(df: pd.DataFrame, chunk_size: int):
    num_chunks = math.ceil(len(df) / chunk_size)
    return [df.iloc[i*chunk_size:(i+1)*chunk_size] for i in range(num_chunks)]

# 6. RUN THE LLAMA EMBEDDING EXTRACTION
output_dir = f"llama_embeddings_output_ds_name_{ds_name}_{flag_case}"
os.makedirs(output_dir, exist_ok=True)

max_workers = 4
batch_size = 8

if embedding_method.lower() == 'llama3':
    for column in embedding_columns:
        print(f"\nProcessing embeddings for column: {column}")

        # Create combined text
        df[f'combined_{column}'] = df.apply(lambda row: create_combined_text(row, column), axis=1)
        texts = df[f'combined_{column}'].tolist()

        # Break texts into batches
        batches = [texts[i:i + batch_size] for i in range(0, len(texts), batch_size)]

        # Generate embeddings in parallel
        embeddings = []
        with concurrent.futures.ThreadPoolExecutor(max_workers=max_workers) as executor:
            futures = [executor.submit(get_llama_embeddings, batch) for batch in batches]
            for future in tqdm(concurrent.futures.as_completed(futures),
                               total=len(futures),
                               desc="Generating Llama Embeddings"):
                embeddings.extend(future.result())

        # Prepare the results in a DataFrame
        embedding_df = pd.DataFrame({
            'itemId': df['itemId'],
            'embeddings': embeddings
        })
        embedding_df['embeddings'] = embedding_df['embeddings'].apply(
            lambda x: ' '.join(map(str, x)) if x else ''
        )

        # Shuffle for random distribution
        embedding_df = embedding_df.sample(frac=1, random_state=42).reset_index(drop=True)

        # Split and save
        chunk_size = estimate_chunk_size(embedding_df, sample_size=1000, max_size_mb=25)
        chunks = split_dataframe(embedding_df, chunk_size)

        for idx, chunk in enumerate(chunks):
            output_file = os.path.join(output_dir, f"llama_{column}_part{idx+1}.csv.gz")
            with gzip.open(output_file, 'wt', encoding='utf-8') as f:
                chunk.to_csv(f, index=False)
            print(f"Saved: {output_file}")

    print("\nLlama Embedding Extraction Complete!")

    # Zip and download the folder
    zip_file_name = f"{output_dir}.zip"
    shutil.make_archive(output_dir, 'zip', output_dir)
    files.download(zip_file_name)
```
